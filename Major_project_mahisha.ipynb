# **4. Data Cleaning and Preprocessing**

Data cleaning is performed to improve the quality and reliability of the
dataset before conducting further analysis.

The dataset is checked for:
- Missing values
- Duplicate records
- Incorrect data types
- Invalid or inconsistent categorical values
- Potential outliers

```python
# Detailed missing-value analysis

missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

# Keep only columns with missing values
missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
]

# Round percentage to 2 decimal places
missing_summary["Missing Percentage"] = missing_summary["Missing Percentage"].round(2)

# Sort by missing count
missing_summary.sort_values(
    by="Missing Count",
    ascending=False
)
```

```python
# Examine missing values by season

missing_by_season = df.groupby("Season")[
    ["Rainfall_mm", "Soil_Moisture_pct", "Yield_Tonnes_Ha"]
].apply(lambda x: x.isnull().sum())

missing_by_season
```

# **6. Outlier Investigation**

Outliers are observations that are unusually high or low compared with the
majority of the data.

Outliers are investigated using the Interquartile Range (IQR) method and
boxplots. They are not automatically removed because extreme agricultural
values may represent genuine differences between farms, crops, seasons, or
production conditions.

```python
# Detect potential outliers using the IQR method

outlier_summary = []

outlier_columns = [
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Total_Cost_INR",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3"
]

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_summary.append({
        "Variable": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier Percentage": (len(outliers) / len(df)) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary["Outlier Percentage"] = \
    outlier_summary["Outlier Percentage"].round(2)

outlier_summary
```

The economic comparison shows an even larger seasonal difference. Kharif has

# **12. Student-Designed Analysis 1: Irrigation Method vs Water Efficiency**

This analysis compares water-use efficiency across different irrigation methods
to identify which irrigation practices provide better efficiency in agricultural
water usage.

```python
# Irrigation Method vs Water Efficiency

irrigation_efficiency = df.groupby(
    "Irrigation_Method"
)["Water_Efficiency_t_per_1000m3"].mean().sort_values(
    ascending=False
)

print("Average Water Efficiency by Irrigation Method:")
print(irrigation_efficiency.round(2))

plt.figure(figsize=(9, 5))

sns.barplot(
    x=irrigation_efficiency.index,
    y=irrigation_efficiency.values
)

plt.title("Average Water Efficiency by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Water Efficiency (t per 1000 m³)")

plt.tight_layout()
plt.show()
```

# **12. Student-Designed Analysis 2: Crop-wise Profit Margin**

This analysis compares profit margins across different crops to identify
which crops provide better economic returns relative to their revenue.

```python
# Calculate Profit Margin

df["Profit_Margin_pct"] = (
    df["Profit_INR"] / df["Revenue_INR"]
) * 100

crop_profit_margin = df.groupby("Crop")["Profit_Margin_pct"].mean().sort_values(
    ascending=False
)

print("Average Profit Margin by Crop:")
print(crop_profit_margin.round(2))

plt.figure(figsize=(10, 6))

sns.barplot(
    x=crop_profit_margin.index,
    y=crop_profit_margin.values
)

plt.title("Average Profit Margin by Crop")
plt.xlabel("Crop")
plt.ylabel("Average Profit Margin (%)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

The crop-wise profit margin shows substantial differences in economic
performance. Chilli has the highest average profit margin, followed by
Sugarcane. Several other crops show negative average profit margins,
indicating that their average costs exceed their average revenue in the
dataset. Rice, Wheat, and Maize show particularly low profit margins,
highlighting substantial variation in profitability across crops.

# **12. Student-Designed Analysis 3: Disease/Pest Risk vs Crop Yield**

This analysis investigates the relationship between disease and pest risk and
crop yield to determine whether higher risk levels are associated with changes
in agricultural productivity.

```python
# Disease/Pest Risk vs Crop Yield

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Disease_Pest_Risk_pct",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.6
)

plt.title("Disease/Pest Risk vs Crop Yield")
plt.xlabel("Disease/Pest Risk (%)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.legend(title="Season")

plt.tight_layout()
plt.show()
```

```python
# Correlation between Disease/Pest Risk and Yield

disease_yield_corr = df["Disease_Pest_Risk_pct"].corr(
    df["Yield_Tonnes_Ha"]
)

print(
    f"Correlation between Disease/Pest Risk and Yield: "
    f"{disease_yield_corr:.3f}"
)
```

### Observation

The scatter plot shows the relationship between disease/pest risk and crop
yield across the three seasons. The calculated correlation coefficient is
0.013, indicating an extremely weak positive linear relationship between
disease/pest risk and yield in the dataset.

Therefore, disease/pest risk alone does not appear to have a meaningful
linear association with crop yield based on this analysis. Other factors
may have a stronger influence on agricultural productivity.
